# Create New Training Data from All_images Folder

Now we'll create a new training dataset using the relationships_updated.csv and matching images from the All_images folder.

```python
# Load the relationships_updated.csv file
relationships_df = pd.read_csv("../data_csv/relationships_updated.csv")
print(f"Relationships data shape: {relationships_df.shape}")
print(f"\nColumns: {relationships_df.columns.tolist()}")
print(f"\nFirst few rows:")
relationships_df.head(10)
```

```python
# Create a mapping of all available images in All_images folder
# img_files already contains the list of images from All_images
image_lookup = {}
for img_path in img_files:
    # Extract just the filename (e.g., "X.010.A.01-1.jpg")
    filename = os.path.basename(img_path)
    image_lookup[filename] = img_path

print(f"Total images found in All_images: {len(image_lookup)}")
print(f"\nSample image mappings:")
for i, (filename, path) in enumerate(list(image_lookup.items())[:5]):
    print(f"{filename} -> {path}")
```

```python
# Load the elements.csv to get all unique elements (these will be our columns)
elements_df = pd.read_csv("../data_csv/elements.csv")
unique_elements = sorted(elements_df['element'].unique())
print(f"Total unique elements: {len(unique_elements)}")
print(f"\nFirst 10 elements: {unique_elements[:10]}")
```

```python
# Group relationships by glyph_image to get all elements for each image
glyph_elements = relationships_df.groupby('glyph_image')['element_name'].apply(list).to_dict()

print(f"Total unique glyph images in relationships: {len(glyph_elements)}")
print(f"\nSample glyph with elements:")
for i, (glyph, elements) in enumerate(list(glyph_elements.items())[:3]):
    print(f"{glyph}: {elements}")
```

```python
# Create the new training data
training_data = []

for glyph_image, element_list in glyph_elements.items():
    # Check if the image exists in All_images folder
    if glyph_image in image_lookup:
        # Create the row with image path and elements
        row = {
            'glyph_image': glyph_image,
            'elements_original': '|'.join(element_list),  # Store as pipe-separated string
            'full_image_path': f"All_images/{image_lookup[glyph_image]}"
        }
        
        # Add binary columns for each element (1 if present, 0 if not)
        for element in unique_elements:
            row[element] = 1 if element in element_list else 0
        
        training_data.append(row)

# Create the DataFrame
new_training_df = pd.DataFrame(training_data)

print(f"New training data shape: {new_training_df.shape}")
print(f"Old training data shape: {df.shape}")
print(f"\nImages found: {len(new_training_df)}")
print(f"Images not found: {len(glyph_elements) - len(new_training_df)}")
```

```python
# Verify the pattern: glyph_cote + "-" + codex_id
# The glyph_image column should already be in this format
print("Checking glyph_image format in relationships_updated.csv:")
print(relationships_df[['glyph_cote', 'codex_id', 'glyph_image']].head(10))

# Verify if the pattern matches
sample_row = relationships_df.iloc[0]
expected_filename = f"{sample_row['glyph_cote']}-{sample_row['codex_id']}.jpg"
print(f"\nExpected: {expected_filename}")
print(f"Actual:   {sample_row['glyph_image']}")
```

```python
# Search for files that might match the pattern
# Let's look for any file containing "X.010" or similar patterns
print("Looking for files with 'X.010' pattern:")
matching_files = [f for f in image_lookup.keys() if 'X' in f or 'x' in f]
print(f"Files with 'X' or 'x': {len(matching_files)}")
if matching_files:
    print(f"Sample: {matching_files[:20]}")
else:
    # Try looking for files with dots
    dot_files = [f for f in list(image_lookup.keys())[:100] if '.' in f]
    print(f"\nFiles with dots: {len(dot_files)}")
    if dot_files:
        print(f"Sample: {dot_files[:20]}")
```

```python
# Create an enhanced lookup that can match by glyph_cote pattern
# Map by both full filename and by glyph_cote pattern
enhanced_lookup = {}

for filename, path in image_lookup.items():
    # Store by exact filename
    enhanced_lookup[filename] = path
    
    # Also create entries for partial matches (e.g., "X.010.A.01" from "X.010.A.01.jpg")
    base_name = filename.rsplit('.', 1)[0]  # Remove extension
    enhanced_lookup[base_name] = path

print(f"Enhanced lookup size: {len(enhanced_lookup)}")

# Test with a sample
test_glyph = "X.010.A.01-1.jpg"
test_cote = "X.010.A.01"
print(f"\nTest lookup for '{test_glyph}': {test_glyph in enhanced_lookup}")
print(f"Test lookup for '{test_cote}': {test_cote in enhanced_lookup}")
if test_cote in enhanced_lookup:
    print(f"  -> {enhanced_lookup[test_cote]}")
```

```python
# Recreate training data using glyph_cote for matching
training_data_new = []

for glyph_image, element_list in glyph_elements.items():
    # Extract glyph_cote from glyph_image (e.g., "X.010.A.01-1.jpg" -> "X.010.A.01")
    # This should match the format in the relationships
    glyph_cote = relationships_df[relationships_df['glyph_image'] == glyph_image]['glyph_cote'].iloc[0] if glyph_image in relationships_df['glyph_image'].values else None
    
    if glyph_cote and glyph_cote in enhanced_lookup:
        # Create the row with image path and elements
        row = {
            'glyph_image': glyph_image,
            'elements_original': '|'.join(element_list),
            'full_image_path': f"All_images/{enhanced_lookup[glyph_cote]}"
        }
        
        # Add binary columns for each element
        for element in unique_elements:
            row[element] = 1 if element in element_list else 0
        
        training_data_new.append(row)

# Create the DataFrame
new_training_df = pd.DataFrame(training_data_new)

print(f"✓ New training data shape: {new_training_df.shape}")
print(f"  Old training data shape: {df.shape}")
print(f"\n✓ Images found: {len(new_training_df)}")
print(f"✗ Images not found: {len(glyph_elements) - len(new_training_df)}")
```
